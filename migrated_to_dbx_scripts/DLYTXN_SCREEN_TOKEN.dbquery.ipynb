```sql
-- MULTISET is implicit in Delta; the source database qualifier is omitted because no target catalog mapping was supplied.
-- Teradata FALLBACK, journals, CHECKSUM, MERGEBLOCKRATIO, and MAP storage clauses have no Databricks SQL equivalent and were removed.
-- CHARACTER SET LATIN was removed because Databricks uses UTF-8; NOT CASESPECIFIC is preserved with UTF8_LCASE collation.
-- TODO: manual review required — confirm bounded VARCHAR/CHAR and fixed-width CHAR trailing-space behavior meet downstream contracts.
-- The Teradata non-unique primary index is replaced with liquid clustering on the same access keys.
CREATE TABLE DLYTXN_SCREEN_TOKEN (
    TXN_REF VARCHAR(35) COLLATE UTF8_LCASE NOT NULL,
    SUPPLIER_CD CHAR(4) COLLATE UTF8_LCASE NOT NULL,
    NM_ROLE_CD CHAR(2) COLLATE UTF8_LCASE NOT NULL,
    TOKEN_SEQ_NR SMALLINT NOT NULL,
    TOKEN_TXT VARCHAR(60) COLLATE UTF8_LCASE
)
CLUSTER BY (TXN_REF, SUPPLIER_CD);
```
