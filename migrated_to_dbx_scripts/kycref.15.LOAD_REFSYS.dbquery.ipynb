```sql
/**************************************************************************\
* CLEAR THE PRIMARY WORKTABLE PRIOR TO LOADING                             *
\**************************************************************************/

-- Converted Teradata two-part database object to the current Unity Catalog namespace.
DELETE FROM aml_refsys_raw;


/**************************************************************************\
* LOAD THE FIXED-WIDTH REFERENCE EXTRACT                                   *
* ------------------------------------------------------------------------ *
* REFSYS daily reference extract (fixed width, 253 bytes)                  *
* A = record type, B = business key, C = payload                           *
* Header (HD) and trailer (TR) records are filtered from the load           *
\**************************************************************************/

-- TODO: manual review required — MultiLoad replacement must confirm the source path and job wrapper.
-- Converted the MultiLoad fixed-width layout and parameterized INSERT to a Databricks read_files projection.
-- Converted quoted Teradata identifiers to legal unquoted Databricks identifiers.
INSERT INTO aml_refsys_raw
(
    a,
    b,
    c
)
SELECT
    SUBSTRING(value, 1, 2) AS a,
    SUBSTRING(value, 3, 30) AS b,
    SUBSTRING(value, 33, 221) AS c
FROM read_files('../data/kycref.refsys.dat', format => 'text')
WHERE SUBSTRING(value, 1, 2) NOT IN ('HD', 'TR');
```
